## 2. Target or proxy

### **Target Definition:** Observed Content Decline (`target_is_declining`)

**Target Details & Origin:**
* **Label Variable:** Binary indicator derived from observed performance data (`1` = traffic/impression decline, `0` = stable or growing).
* **Source:** Derived from the `trend_direction` field in the dataset (specifically where `trend_direction == 'down'`), backed by continuous traffic signals like `impressions_last_30d` versus `impressions_prev_30d`.
* **Nature of Label:** This target is an **observed historical outcome** measured in the performance metrics, not an arbitrary rule or subjective rating. Predicting actual observed performance drops ensures that the model learns genuine traffic decay patterns rather than echoing human assumptions.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load dataset
df = pd.read_csv("https://raw.githubusercontent.com/smag-ev/flyrank-tasks/main/data/raw/content_refresh_anonymized.csv")

# Derive binary target column based on observed traffic trend
df['is_declining'] = (df['trend_direction'] == 'down').astype(int)

# Check target base rate
declining_count = df['is_declining'].sum()
total_count = len(df)
print(f"Total Articles: {total_count:,}")
print(f"Declining Target Count (1s): {declining_count:,} ({declining_count/total_count:.1%})")
```

## 3. Success metric

### **Success Metric:** Precision@K (Primary: `Precision@50`)

**Why this metric:**
In an operational workflow, content teams operate under bandwidth constraints and can only review a fixed number of articles per sprint (e.g., $K = 50$ articles). Standard accuracy or ROC-AUC metrics evaluate model performance across the entire dataset, but editorial teams only care about the accuracy of the top recommendations in their review queue.

`Precision@50` directly measures business utility: *"Of the top 50 articles flagged by the model, what proportion actually suffered true content decline?"*

**Performance Benchmark:**
* **Baseline Rate:** ~54.2% (random or unoptimized distribution of declining articles in the raw dataset).
* **Success Target:** Achieving **> 85.0% Precision@50**, ensuring that team members rarely waste operational capacity reviewing healthy, high-performing content.

## 5. Why ML beats a fixed rule here

### **Why Machine Learning Beats a Fixed Rule:**

1. **Non-linear Multi-Factor Signals:** A simple rule such as *"Flag if age > 180 days"* creates massive false positives because many older articles continue to generate stable traffic. Conversely, newer articles can suffer rapid position decay due to intent shifts.
2. **Complex Feature Interactions:** Content decay is a tangled interaction of multiple features happening simultaneously—such as subtle drops in impression volume (`impressions_last_30d` vs `impressions_prev_30d`), search volume competition, average ranking position drops, and lower engagement rates. ML captures these overlapping signals far better than rigid `if-else` logic.
3. **Continuous Decision-Support Prioritization:** Fixed rules yield binary flags without relative sorting. Machine learning assigns a granular, dynamic score to every article, automatically adjusting the review queue as new performance signals enter the pipeline.
