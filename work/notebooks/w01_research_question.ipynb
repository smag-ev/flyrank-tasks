## 2. The question: decision, action, cost of a wrong call

* **Decision:** Which published web pages should the editorial team review and update first to prevent loss of organic search traffic?
* **Who acts:** Content managers, SEO strategists, and editorial copywriters.
* **Action:** Reviewing target pages to update outdated facts, expand thin sections, optimize search intent match, or flag for consolidation/monitoring.
* **Cost of a wrong call:**
  * *False Positive (recommending a healthy page):* Wastes limited editorial capacity updating content that did not need intervention.
  * *False Negative (missing a decaying page):* Allows valuable, high-traffic pages to quietly lose search rankings and impressions to competitors without timely action.
* **Why ML helps:** Machine learning can simultaneously weigh multiple subtle historical signals—such as ranking position decay, impression trends, content age, and engagement drops—to prioritize true decay patterns better than rigid manual rules.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load starter dataset
df = pd.read_csv("https://raw.githubusercontent.com/smag-ev/flyrank-tasks/main/data/raw/content_refresh_anonymized.csv")

# Extract 3 supporting metrics
total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
declining_pct = (declining_pages / total_pages) * 100
avg_90d_impressions = df["impressions_90d"].mean()
stale_declining_pages = ((df["trend_direction"] == "down") & (df["content_age_days"] >= 180)).sum()

print(f"Total articles in dataset: {total_pages:,}")
print(f"Articles in declining trend: {declining_pages:,} ({declining_pct:.1f}%)")
print(f"Average 90-day impressions per article: {avg_90d_impressions:,.1f}")
print(f"Stale articles (>= 180 days old) currently declining: {stale_declining_pages:,}")
```
